## 💡 Diagnóstico y Plan de Acción (Insights del EDA)

Tras analizar el reporte generado por mi módulo, he detectado varios problemas críticos de calidad de datos que deberé solucionar en la fase de ETL (Siguiente Notebook). Esta es mi hoja de ruta:

### 🚨 1. Problemas de Formato (Tipos de datos incorrectos)
* **Números guardados como texto (Object):** Columnas como `cons.price.idx`, `cons.conf.idx`, `euribor3m` y `nr.employed` deberían ser numéricas (float). Se han guardado como texto porque **usan una coma decimal (`,`) en lugar de un punto (`.`)**. Tendré que reemplazar esto y convertirlas.
* **Fechas como texto:** La columna `date` contiene texto mezclando números y meses en español (ej. *18-diciembre-2015*). Necesita ser parseada a formato `datetime`.

### 🚨 2. Inconsistencias en Texto (Mayúsculas y minúsculas)
* Al observar las frecuencias, veo que `marital` está todo en MAYÚSCULAS ('MARRIED'), mientras que `job` está en minúsculas ('admin.'). 
* Hay caracteres extraños y puntuación en variables categóricas (ej. `admin.`, `blue-collar`). 
* **Plan:** Estandarizaré todo el dataset a minúsculas y reemplazaré guiones y puntos por barras bajas (`_`) para tener consistencia.

### 🚨 3. Gestión de Nulos
Tenemos una cantidad significativa de valores perdidos:
* `euribor3m` (~21.5%) y `default` (~20.8%) tienen un gran volumen de nulos.
* `age` (~11.9%) también falta bastante.
* **Plan:** En el próximo notebook evaluaré si imputar estos valores (con la mediana o moda) o si crear una categoría 'desconocido'.

### 🚨 4. Estandarización de Columnas
* La columna `Unnamed: 0` parece ser un índice residual del CSV, la eliminaré.
* Los nombres de las columnas usan puntos (`emp.var.rate`). Aplicaré una función de mi módulo de limpieza para pasarlos a formato `snake_case` (`emp_var_rate`).
