# 🔢 Análisis Descriptivo de los Datos

## 🎯 ¿Qué busco aquí?
Mi objetivo es hacer una **radiografía estadística** a la tabla que ya limpié. Antes de dibujar gráficos, quiero entender la **naturaleza numérica** de los clientes: ¿Qué edad media tienen? ¿Cuánto ganan? ¿Cuáles son sus perfiles más comunes?

## 📂 Mis datos listos
* **Archivo:** Uso mi archivo final limpio: `../data/procesados/datos_finales_analisis.csv`
* **Contenido:** Unos 20.000 registros donde crucé los datos del banco con los de los clientes.

## 🔎 ¿Qué métricas voy a analizar?
1.  **Tendencia Central:** Quiero ver cuál es el perfil "promedio" del cliente (Media y Mediana).
2.  **Dispersión:** Para ver si hay mucha diferencia entre clientes (Desviación estándar).
3.  **Frecuencias (Moda):**
    * **Target (`y`):** Ver de base cuál es la tasa de éxito de las llamadas (Sí vs. No).
    * **Perfil:** ¿Qué trabajos, estados civiles y estudios se repiten más?

## 🛠️ Herramientas
* **Pandas:** Usaré funciones estadísticas nativas como `describe`, `mode` y `value_counts`.

```python
# -------------------------------------------------------------------------
# CÓDIGO: EXTRACCIÓN DE ESTADÍSTICAS
# -------------------------------------------------------------------------
import pandas as pd

# 1. CARGA DE DATOS
# Cargo el archivo ya limpio aquí para tener el 'df' disponible.
df = pd.read_csv('../data/procesados/datos_finales_analisis.csv')

# 2. Resumen Numérico. Uso la T de Transpuesta para que se lea mejor.
print("--- 📊 ESTADÍSTICAS VARIABLES NUMÉRICAS ---")
display(df.describe().T)

# 3. Análisis del Target. Saco números absolutos y el porcentaje.
print("\n--- 🎯 BALANCE DE LA CAMPAÑA (Target) ---")
conteo = df['y'].value_counts()
porcentaje = df['y'].value_counts(normalize=True) * 100

# Construyo un pequeño dataframe para que el profesor lo vea como una tabla bonita.
df_target = pd.DataFrame({'Total': conteo, 'Porcentaje (%)': porcentaje})
display(df_target)

# 4. Perfil Típico. Extraigo las columnas de texto para sacar su moda (valor más repetido).
print("\n--- 📋 PERFIL TÍPICO (Moda) ---")
cols_cat = df.select_dtypes(include='object').columns

for col in cols_cat:
    # Verifico que la columna exista por seguridad y saco la moda.
    if col in df.columns:
        moda = df[col].mode()[0]
        print(f"-> {col.upper()}: Lo más común es '{moda}'")
```
