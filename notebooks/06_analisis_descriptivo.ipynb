# 📊 06. Análisis Descriptivo de los Datos

## 🎯 Objetivo de este Notebook
Realizar una radiografía estadística a nuestra **Tabla Maestra** (Capa Gold). Antes de generar representaciones gráficas complejas, es fundamental entender matemáticamente cómo se distribuyen nuestros datos y perfilar a nuestra base de clientes.

**Métricas a analizar:**
1. **Tendencia Central y Dispersión:** Media, Mediana, Desviación Estándar y Cuartiles de las variables numéricas.
2. **Balance de Clases (Target):** Distribución de éxito en la campaña (`y`). Fundamental para entender si el dataset está desbalanceado.
3. **Perfil Típico (Moda):** Construcción del cliente "promedio" evaluando las categorías más frecuentes.


```python
# -------------------------------------------------------------------------
# 1. CONFIGURACIÓN Y CARGA DE DATOS
# -------------------------------------------------------------------------
import pandas as pd

# Configuración visual de Pandas para analítica
pd.set_option('display.max_columns', None)
pd.options.display.float_format = '{:.2f}'.format # Restringe a 2 decimales y evita notación científica

# Cargar la tabla maestra procesada
ruta_maestra = '../data/procesados/datos_finales_analisis.csv'
df = pd.read_csv(ruta_maestra)

print(f"✅ Datos cargados correctamente. Dimensiones de la tabla: {df.shape}")
```

## 🔢 Paso 1: Resumen Numérico
Usamos estadística descriptiva básica para entender las edades, duraciones e indicadores económicos. Empleamos la matriz transpuesta (`.T`) para facilitar la lectura vertical.


```python
# -------------------------------------------------------------------------
# 2. ESTADÍSTICAS VARIABLES NUMÉRICAS
# -------------------------------------------------------------------------
print("--- 📊 ESTADÍSTICAS DESCRIPTIVAS ---")

# Generamos resumen estadístico transpuesto
stats_num = df.describe().T
display(stats_num)

print("\n💡 OBSERVACIONES CLAVE:")
print("- Varianza Cero: Las variables 'default', 'pdays' y 'previous' presentan desviación estándar (std) de 0.00. No aportan poder predictivo en esta muestra.")
print("- Nulos Masivos: La columna 'cons_conf_idx' presenta count=0 (vacía).")
print("- Edades: Nuestro público oscila entre los 19 y 61 años, con una edad media de casi 40 años.")
```

## 🎯 Paso 2: Análisis de la Variable Objetivo (Target)
Nuestra variable a predecir o entender es `y` (si el cliente contrató el producto o no). Es vital conocer la tasa base de éxito de la campaña.


```python
# -------------------------------------------------------------------------
# 3. BALANCE DE LA CAMPAÑA (Target)
# -------------------------------------------------------------------------
print("--- 🎯 DISTRIBUCIÓN DEL TARGET ('y') ---")

# Cálculo de absolutos y relativos
conteo = df['y'].value_counts()
porcentaje = df['y'].value_counts(normalize=True) * 100

# Construcción de un DataFrame de presentación
df_target = pd.DataFrame({
    'Total Llamadas': conteo, 
    'Porcentaje': porcentaje.round(2).astype(str) + ' %'
})

display(df_target)

print("\n💡 Conclusión: Estamos ante un dataset fuertemente desbalanceado (~95.4% No vs ~4.6% Sí).")
print("Esto significa que la campaña tiene una tasa de conversión baja y el perfilado será vital para optimizar futuras llamadas.")
```

## 📋 Paso 3: El Perfil Típico (La Moda)
Vamos a extraer el valor más repetido de nuestras variables categóricas (texto) para construir un perfil sociodemográfico de nuestro "cliente promedio". Filtramos fechas e IDs, ya que no aportan valor a la segmentación general.


```python
# -------------------------------------------------------------------------
# 4. PERFIL TÍPICO DE CLIENTE (CATEGÓRICAS)
# -------------------------------------------------------------------------
print("--- 👤 PERFIL DEL CLIENTE MÁS COMÚN ---")

# 1. Filtramos columnas: Solo texto, ignorando IDs y fechas
cols_ignorar = ['ID', 'date', 'dt_customer', 'y']
cols_cat = [col for col in df.select_dtypes(include='object').columns if col not in cols_ignorar]

# 2. Extraemos la moda iterando
perfil = []
for col in cols_cat:
    moda = df[col].mode()[0]
    perfil.append({'Atributo Demográfico': col.upper(), 'Valor más habitual (Moda)': moda})

# 3. Mostramos en tabla
df_perfil = pd.DataFrame(perfil)
display(df_perfil)
```

## 🧹 Paso 4: Limpieza Post-Análisis (Feature Selection Básica)
Gracias al resumen estadístico, hemos detectado columnas que no aportan varianza o están vacías. Vamos a eliminarlas del dataset para no arrastrar "ruido" a nuestra fase de visualización y modelado.


```python
# -------------------------------------------------------------------------
# 5. ELIMINACIÓN DE COLUMNAS SIN VARIANZA / NULAS
# -------------------------------------------------------------------------
# 'default', 'pdays', 'previous' -> Tienen desviación estándar 0 (son constantes)
# 'cons_conf_idx' -> Está 100% vacía
columnas_basura = ['default', 'pdays', 'previous', 'cons_conf_idx']

# Eliminamos las columnas (errors='ignore' evita fallos si ya se borraron antes)
df.drop(columns=columnas_basura, inplace=True, errors='ignore')

print(f"🗑️ Se han eliminado {len(columnas_basura)} columnas sin valor analítico.")
print(f"📊 Nuevas dimensiones del dataset: {df.shape}")

# Sobrescribimos el archivo maestro para que el Notebook 07 ya lo lea limpio
df.to_csv('../data/procesados/datos_finales_analisis.csv', index=False)
print("💾 Archivo maestro actualizado con éxito.")
```
